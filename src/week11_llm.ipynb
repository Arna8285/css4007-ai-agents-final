```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI

# Загружаем обновлённый API-ключ
load_dotenv("../.env", override=True)

# Создаём модель
llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)

print("Model initialized successfully!")
```

```python
# Отправляем первый финансовый запрос
try:
    response = llm.invoke(
        "My monthly income is 250000 KZT. "
        "My expenses are 180000 KZT. "
        "How much money can I save?"
    )

    print("AI response:")
    print(response.content)

except Exception as error:
    print("Error type:", type(error).__name__)
    print("The request failed. Check API authentication and billing.")
```

```python
from langchain_core.messages import SystemMessage, HumanMessage

# Определяем роль финансового помощника
system_message = SystemMessage(
    content=(
        "You are an AI Personal Finance Agent. "
        "Help users understand their income, expenses, "
        "monthly budgets, and savings goals. "
        "Give clear and simple answers. "
        "Use KZT as the currency. "
        "Do not invent financial information."
    )
)

# Создаём вопрос пользователя
user_message = HumanMessage(
    content=(
        "My income is 300000 KZT. "
        "My expenses are 220000 KZT. "
        "How much can I save in 6 months?"
    )
)

# Отправляем инструкции и вопрос модели
response = llm.invoke([system_message, user_message])

# Показываем ответ
print(response.content)
```

```python
# Проверяем финансовую цель
user_message = HumanMessage(
    content=(
        "My monthly income is 200000 KZT. "
        "My monthly expenses are 170000 KZT. "
        "I want to save 500000 KZT in 6 months. "
        "Is my goal realistic? Explain briefly."
    )
)

# Отправляем запрос модели
response = llm.invoke([system_message, user_message])

# Показываем ответ
print(response.content)
```
